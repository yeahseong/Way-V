학습·테스트 폴더의 클래스별 이미지 개수를 세어 구성이 의도대로 되었는지 확인한 노트북.

8만 장 규모를 다루다 보니 Drive 동기화 지연, 경로 불일치, 유니코드 파일명 문제로
개수가 어긋나는 일이 반복됐다. 학습을 돌리기 전에 **클래스 불균형과 누락을 먼저 잡기 위해**
매번 이 검증을 거쳤다.


```python
from google.colab import drive
import os

print("🔗 구글 드라이브 연결을 시작합니다...")
# 1. 구글 드라이브 마운트 (팝업창이 뜨면 '허용'을 눌러주세요)
drive.mount('/content/drive')
print("-" * 40)

# 2. 최상위 테스트 폴더 경로 지정
base_path = '/content/drive/MyDrive/WayV_Detection_Test'

print("🔍 폴더별 사진 개수를 세는 중입니다...\n")

if os.path.exists(base_path):
    total_images = 0
    # base_path 안에 있는 폴더들을 하나씩 확인
    for model_folder in sorted(os.listdir(base_path)):
        model_path = os.path.join(base_path, model_folder)

        # 파일이 아닌 '폴더'인 경우에만 개수 세기 진행
        if os.path.isdir(model_path):
            image_count = 0

            # os.walk가 하위 폴더(0_real, 1_fake 등)를 끝까지 파고들며 탐색
            for root, dirs, files in os.walk(model_path):
                for file in files:
                    # png, jpg, jpeg 등 이미지 파일만 카운트
                    if file.lower().endswith(('.png', '.jpg', '.jpeg')):
                        image_count += 1

            # 결과 출력
            print(f"📂 '{model_folder}' 폴더에는 {image_count:,}장의 사진이 들어있습니다.")
            total_images += image_count

    print("-" * 40)
    print(f"✨ 완료! 12개 모델 폴더에 총 {total_images:,}장의 사진이 있습니다.")
else:
    print(f"❌ 경로를 찾을 수 없습니다. 폴더 이름이나 위치가 맞는지 확인해주세요:\n{base_path}")
```

```python
from google.colab import drive
import os

print("🔗 구글 드라이브 연결을 시작합니다...")
# 구글 드라이브 마운트
drive.mount('/content/drive')
print("-" * 50)

# 최상위 테스트 폴더 경로 지정
base_path = '/content/drive/MyDrive/WayV_Detection_Test'

print("🔍 각 모델별 0_real 및 1_fake 폴더 사진 개수를 세는 중입니다...\n")

if os.path.exists(base_path):
    # base_path 안에 있는 폴더들을 하나씩 확인 (AttGAN, BEGAN 등)
    for model_folder in sorted(os.listdir(base_path)):
        model_path = os.path.join(base_path, model_folder)

        # 파일이 아닌 '폴더'인 경우에만 진행
        if os.path.isdir(model_path):
            real_count = 0
            fake_count = 0

            # os.walk로 폴더 끝까지 탐색
            for root, dirs, files in os.walk(model_path):
                # 현재 탐색 중인 맨 끝단 폴더의 이름 확인
                current_folder_name = os.path.basename(root)

                # 방 이름이 '0_real'인 경우
                if current_folder_name == '0_real':
                    for file in files:
                        if file.lower().endswith(('.png', '.jpg', '.jpeg')):
                            real_count += 1

                # 방 이름이 '1_fake'인 경우
                elif current_folder_name == '1_fake':
                    for file in files:
                        if file.lower().endswith(('.png', '.jpg', '.jpeg')):
                            fake_count += 1

            # 둘 중 하나라도 사진이 들어있는 폴더만 출력 (빈 폴더 패스)
            if real_count > 0 or fake_count > 0:
                print(f"'{model_folder}' 폴더에 0_real 폴더에는 {real_count:,}장의 사진이 있습니다. 1_fake 폴더에는 {fake_count:,}장의 사진이 있습니다.")

    print("\n✨ 모든 폴더의 상세 개수 카운트가 완료되었습니다!")
else:
    print(f"❌ 경로를 찾을 수 없습니다. 폴더 이름이나 위치가 맞는지 확인해주세요:\n{base_path}")
```

```python
import os

target_path = '/content/drive/MyDrive/WayV_Detection_Train/train/1_fake/StyleGAN2'

print("⚔️ 유니코드 버그를 회피하는 무적의 절단기를 가동합니다...\n")

count = 0
if os.path.exists(target_path):
    for filename in os.listdir(target_path):
        # 정상적인 이미지 확장자(.png)로 끝나지 않는 362개의 파일만 골라냅니다.
        if not filename.lower().endswith(('.png', '.jpg', '.jpeg')):

            # 이름 중간에 '.png'라는 글자가 있다면?
            if ".png" in filename.lower():
                # '.png'를 기준으로 문자를 쪼갠 뒤, 앞부분만 남기고 다시 .png를 붙입니다.
                new_filename = filename.split(".png")[0] + ".png"

                old_file = os.path.join(target_path, filename)
                new_file = os.path.join(target_path, new_filename)

                # 이름 변경!
                os.rename(old_file, new_file)
                count += 1

    print(f"✨ 총 {count}개의 파일 꼬리표를 강제로 잘라내어 복구했습니다!\n")

    # 최종 팩트 체크 (2만 장 맞는지 확인)
    final_files = [f for f in os.listdir(target_path) if f.lower().endswith('.png')]
    print("-" * 50)
    print(f"🎉 팩트 체크: 현재 [StyleGAN2] 폴더 안에는 총 **{len(final_files):,}장**의 사진이 있습니다.")
    print("-" * 50)

else:
    print("❌ 경로를 찾을 수 없습니다.")
```
